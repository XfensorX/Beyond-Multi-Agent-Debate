```python
from typing import Literal
import polars as pl

from social_groups.analysis.defs.notebooks.definitions import register_materialization
from social_groups.analysis.polars_transformations import (
    apply_parsing_and_group_decision,
)
from social_groups.analysis.polars_transformations.wasserstein_distance import (
    calculate_wasserstein_distance,
)
from social_groups.reporting.group_reply import (
    GroupReplyAggregator,
    MajorityVote,
)
from social_groups.reporting.parsing import (
    AnswerComparer,
    AnswerOptions,
    AnswerParser,
)

import social_groups.polars_columns as plc

%load_ext autoreload
%autoreload 2
```

```python
parser = AnswerParser(AnswerOptions.letters_A_to_J)
group_reply = GroupReplyAggregator(MajorityVote())
comparer = AnswerComparer(
    AnswerOptions.letters_A_to_J, triple_underscore_handling="wrong"
)
```

```python
from social_groups.analysis.definitions import defs

diversity_mad_frame: pl.DataFrame = defs().load_asset_value("final_diversity_params")
```

```python
diversity_mad = apply_parsing_and_group_decision(
    diversity_mad_frame, parser, comparer, group_reply
)
```

```python
diversity_mad.head()
```

```python
NUM_ROUNDS = diversity_mad["number_of_mad_rounds"].unique().item()
```

```python
diversity_results = (
    diversity_mad.group_by(
        "temperature",
        "top_p",
        "model_family",
        "model_parameters",
        "sizes",
        "model_names",
    )
    .agg(
        pl.col("run_id").unique().item(),  # To assert correct aggregation
        pl.len(),
        pl.col(plc.is_correct).mean().alias(plc.accuracy),
        pl.col(plc.used_input_tokens).mean(),
        pl.col(plc.used_output_tokens).mean(),
        (
            pl.col("___parsed_individual_answers_after___").list.count_matches(
                "___not_parsable___"
            )
            / pl.col("___parsed_individual_answers_after___").list.len()
        )
        .mean()
        .alias("amount_parsing_errors_after"),
        (
            pl.col("___parsed_individual_answers_before___").list.count_matches(
                "___not_parsable___"
            )
            / pl.col("___parsed_individual_answers_before___").list.len()
        )
        .mean()
        .alias("amount_parsing_errors_before"),
        comparer(
            pl.col(plc.parsed_combined_answers_before),
            pl.col(plc.answer_string),
        )
        .mean()
        .alias("accuracy_before"),
    )
    .sort(plc.accuracy, descending=True)
    .with_columns(
        call_cost_heuristic=pl.col("model_parameters")
        .list.eval(pl.element() * 2 * NUM_ROUNDS)
        .list.sum(),
        accuracy_increase=pl.col("accuracy") - pl.col("accuracy_before"),
        group_size=pl.col("sizes").list.len(),
    )
    .with_columns(
        pl.col("top_p").round(2).cast(pl.String).cast(pl.Float64),
        pl.col("temperature").round(2).cast(pl.String).cast(pl.Float64),
    )
)

diversity_results = diversity_results.join(
    calculate_wasserstein_distance(
        diversity_mad, group_reply, comparer, "run_id"
    ).select("run_id", pl.selectors.starts_with("wd")),
    on="run_id",
    validate="1:1",
).drop("run_id")
```

```python
import math
import polars as pl
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# =========================
# Choose metric here
# =========================
METRIC = "accuracy"  # e.g. "accuracy", "accuracy_increase", "call_cost_heuristic"
AGG_FN = "item"  # "mean", "median", "max", "min"

# =========================
# Settings
# =========================
FAMILY_COL = "model_family"
SIZE_COL = "sizes"
TEMP_COL = "temperature"
TOPP_COL = "top_p"

# If multiple rows exist per cell, this controls aggregation

# =========================
# Prepare dataframe
# =========================
df_plot = diversity_results.with_columns(
    pl.col(SIZE_COL).list.join("").alias("sizes_str")
)

if AGG_FN == "mean":
    agg_expr = pl.col(METRIC).mean()
elif AGG_FN == "median":
    agg_expr = pl.col(METRIC).median()
elif AGG_FN == "max":
    agg_expr = pl.col(METRIC).max()
elif AGG_FN == "min":
    agg_expr = pl.col(METRIC).min()
elif AGG_FN == "item":
    agg_expr = pl.col(METRIC).item()
else:
    raise ValueError("AGG_FN must be one of: mean, median, max, min")

heatmap_df = (
    df_plot.group_by([FAMILY_COL, "sizes_str", TEMP_COL, TOPP_COL])
    .agg(agg_expr.alias(METRIC))
    .sort([FAMILY_COL, "sizes_str", TEMP_COL, TOPP_COL])
    .to_pandas()
)
baseline_df = heatmap_df.query(f"{TEMP_COL} == 0 and {TOPP_COL} == 1")[
    [FAMILY_COL, "sizes_str", METRIC]
].rename(columns={METRIC: "baseline"})

heatmap_df = heatmap_df.merge(baseline_df, on=[FAMILY_COL, "sizes_str"], how="left")

heatmap_df[f"{METRIC}_delta"] = heatmap_df[METRIC] - heatmap_df["baseline"]

PLOT_METRIC = f"{METRIC}_delta"

# =========================
# Plot overview heatmaps
# =========================
groups = (
    heatmap_df[[FAMILY_COL, "sizes_str"]]
    .drop_duplicates()
    .sort_values([FAMILY_COL, "sizes_str"])
)

n_groups = len(groups)
ncols = diversity_results[FAMILY_COL].unique().len()
nrows = math.ceil(n_groups / ncols)

fig, axes = plt.subplots(
    nrows=nrows, ncols=ncols, figsize=(4.2 * ncols, 3.4 * nrows), squeeze=False
)

vmin = heatmap_df[PLOT_METRIC].min()
vmax = heatmap_df[PLOT_METRIC].max()
family_order = heatmap_df[FAMILY_COL].drop_duplicates().sort_values().tolist()

size_order = heatmap_df["sizes_str"].drop_duplicates().sort_values().tolist()

ncols = len(family_order)
nrows = len(size_order)

fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(4.2 * ncols, 3.4 * nrows))

vmin = heatmap_df[PLOT_METRIC].min()
vmax = heatmap_df[PLOT_METRIC].max()

for row_idx, size in enumerate(size_order):
    for col_idx, family in enumerate(family_order):
        ax = axes[row_idx, col_idx]

        sub = heatmap_df[
            (heatmap_df[FAMILY_COL] == family) & (heatmap_df["sizes_str"] == size)
        ]

        if sub.empty:
            ax.axis("off")
            ax.set_title(f"{family} | {size}\nmissing")
            continue

        pivot = (
            sub.pivot(index=TEMP_COL, columns=TOPP_COL, values=PLOT_METRIC)
            .sort_index()
            .sort_index(axis=1)
        )

        sns.heatmap(
            pivot,
            ax=ax,
            cmap="viridis",
            annot=True,
            fmt=".3f",
            vmin=vmin,
            vmax=vmax,
            cbar=False,
        )

        if row_idx == 0:
            ax.set_title(family, fontsize=11)

        if col_idx == 0:
            ax.set_ylabel(f"{size}\ntemperature")
        else:
            ax.set_ylabel("")

        if row_idx == nrows - 1:
            ax.set_xlabel("top_p")
        else:
            ax.set_xlabel("")

norm = plt.Normalize(vmin=vmin, vmax=vmax)
sm = plt.cm.ScalarMappable(cmap="viridis", norm=norm)
sm.set_array([])

fig.colorbar(
    sm, ax=axes, orientation="vertical", fraction=0.015, pad=0.02, label=PLOT_METRIC
)

fig.suptitle(
    f"{PLOT_METRIC}: temperature × top_p heatmaps by model family and size",
    fontsize=14,
    y=1.002,
)

# plt.tight_layout()x
# plt.show()
```

```python
import polars as pl
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# =========================
# SETTINGS
# =========================


def heatmap_baseline_diff_mean_across_family(METRIC):

    FAMILY_COL = "model_family"
    SIZE_COL = "sizes"
    TEMP_COL = "temperature"
    TOPP_COL = "top_p"

    BASELINE_TEMP = 0.0
    BASELINE_TOPP = 1.0

    # =========================
    # PREPARE GROUP LABELS
    # =========================
    df = (
        diversity_results.with_columns(pl.col(SIZE_COL).list.join("").alias("group"))
        .filter(pl.col("group").str.len_chars() == 3)
        .filter(pl.col("model_family") == "Qwen3.5")
        # .with_columns(
        #     pl.col("group")
        #     .str.split("")
        #     .list.eval(
        #         pl.when(pl.element() == "H")
        #         .then(1)
        #         .otherwise(pl.when(pl.element() == "L").then(-1).otherwise(0))
        #     ).list.sum()
        # )
    )

    # =========================
    # COMPUTE BASELINE PER FAMILY+GROUP
    # =========================
    baseline = df.filter(
        (pl.col(TEMP_COL) == BASELINE_TEMP) & (pl.col(TOPP_COL) == BASELINE_TOPP)
    ).select([FAMILY_COL, "group", pl.col(METRIC).alias("baseline")])

    df_delta = df.join(baseline, on=[FAMILY_COL, "group"], how="left").with_columns(
        (pl.col(METRIC) - pl.col("baseline")).alias("delta")
    )

    # =========================
    # AGGREGATE ACROSS FAMILIES
    # =========================
    agg = df_delta.group_by(["group", TEMP_COL, TOPP_COL]).agg(
        pl.col("delta").mean().alias("mean_delta")
    )

    # Create column labels
    agg = agg.with_columns(
        (
            pl.lit("T=")
            + pl.col(TEMP_COL).cast(pl.String)
            + pl.lit("\np=")
            + pl.col(TOPP_COL).cast(pl.String)
        ).alias("setting")
    )

    plot_df = agg.to_pandas()

    # =========================
    # GROUP ORDER
    # =========================
    group_order = (
        agg.select("group")
        .unique()
        # .with_columns(
        # pl.col("group").alias("score")
        # )
        .with_columns(
            pl.col("group")
            .map_elements(
                lambda s: sum({"L": 0, "M": 1, "H": 2}[c] for c in s),
                return_dtype=pl.Int64,
            )
            .alias("score")
        )
        .sort(["score", "group"])
        .get_column("group")
        .to_list()
    )

    pivot = plot_df.pivot(index="group", columns="setting", values="mean_delta")

    # preserve sensible order
    pivot = pivot.reindex(group_order[::-1])

    # =========================
    # PLOT
    # =========================
    abs_max = max(abs(pivot.min().min()), abs(pivot.max().max()))

    plt.figure(figsize=(12, max(4, len(pivot) * 0.4)))

    sns.heatmap(
        pivot,
        annot=True,
        fmt=".3f",
        cmap="RdBu_r",
        center=0,
        vmin=-abs_max,
        vmax=abs_max,
    )

    plt.title(f"Mean Δ{METRIC} relative to T={BASELINE_TEMP}, p={BASELINE_TOPP}")
    plt.xlabel("Temperature / Top-p")
    plt.ylabel("Model Group")
    plt.tight_layout()
    plt.show()
```

```python
import polars as pl
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import TwoSlopeNorm


def heatmap_baseline_diff_by_family_stacked(
    metrics,
    *,
    diversity_results,
    family_col="model_family",
    size_col="sizes",
    temp_col="temperature",
    topp_col="top_p",
    baseline_temp=0.0,
    baseline_topp=1.0,
    figsize_per_family=1.35,
    figsize_per_metric=1.15,
    annotate=True,
    fmt=".3f",
    cmap="RdBu_r",
    method: Literal["mean", "std"] = "mean",
):
    df = diversity_results.with_columns(pl.col(size_col).list.join("").alias("group"))

    pieces = []

    if method == "std":
        method_calc = pl.col("delta").std()
    elif method == "mean":
        method_calc = pl.col("delta").mean()
    else:
        raise NotImplementedError(f"Method {method} not implemented")

    for metric in metrics:
        baseline = df.filter(
            (pl.col(temp_col) == baseline_temp) & (pl.col(topp_col) == baseline_topp)
        ).select(
            [
                family_col,
                "group",
                pl.col(metric).alias("baseline"),
            ]
        )

        agg = (
            df.join(baseline, on=[family_col, "group"], how="left")
            .with_columns((pl.col(metric) - pl.col("baseline")).alias("delta"))
            .group_by([family_col, temp_col, topp_col])
            .agg(method_calc.alias(f"{method}_delta"))
            .with_columns(pl.lit(metric).alias("metric"))
        )

        pieces.append(agg)

    plot_df = pl.concat(pieces).to_pandas()

    metrics = list(metrics)
    families = sorted(plot_df[family_col].unique())
    temps = sorted(plot_df[temp_col].unique())
    topps = sorted(plot_df[topp_col].unique())

    n_rows = len(metrics) * len(temps)
    n_cols = len(families) * len(topps)

    values = np.full((n_rows, n_cols), np.nan)

    for i_m, metric in enumerate(metrics):
        for i_f, family in enumerate(families):
            sub = plot_df[
                (plot_df["metric"] == metric) & (plot_df[family_col] == family)
            ]

            pivot = sub.pivot(
                index=temp_col, columns=topp_col, values=f"{method}_delta"
            ).reindex(index=temps, columns=topps)

            r0 = i_m * len(temps)
            c0 = i_f * len(topps)
            values[r0 : r0 + len(temps), c0 : c0 + len(topps)] = pivot.values

    abs_max = np.nanmax(np.abs(values))
    norm = TwoSlopeNorm(vmin=-abs_max, vcenter=0, vmax=abs_max)

    fig_w = max(6, figsize_per_family * len(families))
    fig_h = max(3, figsize_per_metric * len(metrics))

    fig, ax = plt.subplots(figsize=(fig_w, fig_h), constrained_layout=True)

    im = ax.imshow(values, cmap=cmap, norm=norm, aspect="auto")

    # Thin cell grid
    ax.set_xticks(np.arange(-0.5, n_cols, 1), minor=True)
    ax.set_yticks(np.arange(-0.5, n_rows, 1), minor=True)
    ax.grid(which="minor", color="white", linewidth=0.45)
    ax.tick_params(which="minor", bottom=False, left=False)

    # Black separators between families and metrics
    for k in range(1, len(families)):
        ax.axvline(k * len(topps) - 0.5, color="black", linewidth=0.9)

    for k in range(1, len(metrics)):
        ax.axhline(k * len(temps) - 0.5, color="black", linewidth=0.9)

    # One column-label row: top_p repeated under each family block
    ax.set_xticks(np.arange(n_cols))
    ax.set_xticklabels(
        [str(t) for _family in families for t in topps],
        rotation=45,
        ha="right",
        fontsize=8,
    )

    # One row-label column: temperature values
    ax.set_yticks(np.arange(n_rows))
    ax.set_yticklabels(
        [str(t) for _metric in metrics for t in temps],
        fontsize=8,
    )

    # Family labels above blocks
    for i, family in enumerate(families):
        x = i * len(topps) + (len(topps) - 1) / 2
        ax.text(
            x,
            -1.15,
            family,
            ha="center",
            va="bottom",
            fontsize=10,
            fontweight="bold",
            clip_on=False,
        )

    # Metric labels on the left of row blocks
    for i, metric in enumerate(metrics):
        y = i * len(temps) + (len(temps) - 1) / 2
        ax.text(
            -1.35,
            y,
            metric,
            ha="right",
            va="center",
            fontsize=10,
            fontweight="bold",
            # rotation=90,
            clip_on=False,
        )

    if annotate:
        for r in range(n_rows):
            for c in range(n_cols):
                val = values[r, c]
                if np.isfinite(val):
                    ax.text(
                        c,
                        r,
                        format(val, fmt),
                        ha="center",
                        va="center",
                        fontsize=7,
                    )

    ax.set_xlabel("top_p")
    ax.set_ylabel("temperature")

    ax.set_title(
        f"{method.capitalize()} of metric change relative to T={baseline_temp}, p={baseline_topp}\n"
        "Averaged across all 2 and 3-model groups within each family",
        fontsize=12,
        pad=28,
    )

    cbar = fig.colorbar(im, ax=ax, fraction=0.025, pad=0.02)
    cbar.set_label(f"{method.capitalize()} Δ from baseline")

    plt.show()
```

```python
import polars as pl
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# =========================
# SETTINGS
# =========================
OBJECTIVE = "accuracy"

MAXIMIZE = True

FAMILY_COL = "model_family"
SIZE_COL = "sizes"
TEMP_COL = "temperature"
TOPP_COL = "top_p"

# =========================
# PREPARE GROUP LABELS
# =========================
df = diversity_results.with_columns(pl.col(SIZE_COL).list.join("").alias("group"))

# =========================
# FIND BEST SETTING
# =========================
sort_desc = MAXIMIZE

best = (
    df.sort(by=[FAMILY_COL, "group", OBJECTIVE], descending=[False, False, sort_desc])
    .group_by([FAMILY_COL, "group"])
    .first()
)

# =========================
# AVERAGE BEST SETTINGS
# =========================
summary = (
    best.group_by("group")
    .agg(
        [
            pl.col(TEMP_COL).mean().alias("avg_best_temp"),
            pl.col(TOPP_COL).mean().alias("avg_best_top_p"),
            pl.len().alias("n_families"),
        ]
    )
    .sort("group")
)

# =========================
# GROUP ORDER
# H-heavy groups first, L-heavy groups last
# =========================
group_order = (
    summary.select("group")
    .unique()
    .with_columns(
        pl.col("group")
        .map_elements(
            lambda s: list(
                {"H": 2, "M": 1, "L": 0}[c] for c in sorted(s, reverse=True)
            ),
            return_dtype=pl.List(pl.Int64),
        )
        .alias("sort_key")
    )
    .sort("sort_key", descending=True)
    .get_column("group")
    .to_list()
)

summary_pd = summary.to_pandas()

# =========================
# HEATMAP
# =========================
heatmap_data = summary_pd.set_index("group").reindex(group_order)[
    ["avg_best_temp", "avg_best_top_p"]
]

plt.figure(figsize=(4, max(4, len(heatmap_data) * 0.4)))

sns.heatmap(heatmap_data, annot=True, fmt=".2f", cmap="viridis")

plt.title(f"Average Optimal Decoding Parameters\n(Objective = {OBJECTIVE})")
plt.xlabel("")
plt.ylabel("Model Group")
plt.tight_layout()
plt.show()

# Optional table
print(summary_pd)
```

```python
from social_groups.directories import MULTIRUN_FINAL_RESULTS_DIR
```

```python
for experiment_name in MULTIRUN_FINAL_RESULTS_DIR.iterdir():
    if not experiment_name.is_dir():
        continue
    for run_name in experiment_name.iterdir():
        if not run_name.is_dir():
            continue # multirun.yaml

        for single_configuration in run_name.iterdir():

            if not single_configuration.is_dir():
                continue # multirun.yaml
            with open(single_configuration / "stdout.log") as f:
                for line in f.readlines():
                    if "ERROR" in line:
                        print(f"{run_name}: {line}")
                        break
```
